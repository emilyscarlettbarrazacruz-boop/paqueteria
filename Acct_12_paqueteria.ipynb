```python
# Programa de paquetería

# Pedir datos
peso = int(input("Peso del paquete: "))
distancia = int(input("Distancia en km: "))
fragil = input("¿Es frágil? (si/no): ")

# Revisar peso
if peso <= 5:
    costo = 100
else:
    costo = 200

# Revisar distancia
if distancia > 100:
    costo = costo + 80

# Revisar si es frágil
if fragil == "si":
    costo = costo + 50

# Mostrar resultado
print("Total a pagar:", costo)
```
